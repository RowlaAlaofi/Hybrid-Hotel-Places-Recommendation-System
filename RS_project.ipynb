**Collabrative Filltring**

```python
# Load the hotel data into the Surprise format for collaborative filtering
reader = Reader(rating_scale=(1, 10))  # 1-10 scale for customer ratings
data = Dataset.load_from_df(hotel_data[['Property_id', 'hotel_name', 'Customers_Rating']], reader)

algo = SVD()

trainset = data.build_full_trainset()

# Initialize the SVD algorithm
algo.fit(trainset)


def recommend_hotels(user_id, price_range):
    # Filter hotels by the specified price range
    valid_hotels = hotel_data[(hotel_data['Price'] >= price_range[0]) &
                              (hotel_data['Price'] <= price_range[1])]
    recommendations = []
    for hotel_id in valid_hotels['Property_id'].unique():
        # Predict the user's rating for each hotel within the filtered list
        predicted_rating = algo.predict(user_id, hotel_id).est

        # Append hotel with its predicted rating to the recommendations list
        recommendations.append((hotel_id, predicted_rating))

    # Sort hotels by predicted rating in descending order
    recommendations = sorted(recommendations, key=lambda x: x[1], reverse=True)

    # Retrieve hotel details for the top recommendations
    recommended_hotels = hotel_data[hotel_data['Property_id'].isin([hotel[0] for hotel in recommendations])]

    return recommended_hotels[['hotel_name', 'city', 'Price', 'Customers_Rating','Property_id']]


```

**Content-Based Filtering**

```python
def new_get_recommendations(city,Typology, com_data):
    com_data['combined_features'] = (
        com_data['name of place'].astype(str) + " " +
        com_data['city'].astype(str) + " " +
        com_data['type'].astype(str) + " " +
        com_data['Price'].astype(str) + " " +
        com_data['Customers_Rating'].astype(str) + " " +
        com_data['hotel_name'].astype(str)
    )

    # Explicitly handle NaN values and convert them to empty strings
    com_data['combined_features'] = com_data['combined_features'].fillna('')

    filtered_data = com_data[(com_data['city'] == city)& (com_data['type'] == Typology)]
    #filtered_data = com_data[(com_data['Typology'] == Typology)]


    # If there's no match for the city and type, return a message or empty DataFrame
    if filtered_data.empty:
        print("No matches found for the specified city and type.")
        return pd.DataFrame(columns=['name of place', 'city', 'Typology'])

    # Initialize TfidfVectorizer with stop_words=None
    # to prevent automatic removal of words based on a standard list
    vectorizer = TfidfVectorizer(stop_words=None)

    # Vectorize the combined features of only the filtered data
    tfidf_matrix_filtered = vectorizer.fit_transform(filtered_data['combined_features'])

    # Calculate cosine similarity within the filtered subset
    cosine_sim_filtered = cosine_similarity(tfidf_matrix_filtered, tfidf_matrix_filtered)

    # Get the index of the first item in the filtered set for comparison
    idx = filtered_data.index[0]

    # Get similarity scores for the first item in the filtered subset
    sim_scores = list(enumerate(cosine_sim_filtered[0]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Get the indices of the most similar items (excluding the first one, which is itself)
    sim_indices = [filtered_data.index[i[0]] for i in sim_scores[1:6]]  # Top 5 similar places

    # Return the top recommended places
    return com_data.loc[sim_indices, ['name of place', 'city', 'type', 'Price', 'Customers_Rating', 'hotel_name']]
```

Hybrid recommendation

```python
def new_hybrid_recommendations(user_id, com_data,typo, city, price_range):
    # Get collaborative recommendations based on user ID and price range
    collaborative_recommendations = recommend_hotels(user_id, price_range)

    # Ensure 'Property_id' is present in collaborative_recommendations
    if 'Property_id' not in collaborative_recommendations.columns:
        raise ValueError("collaborative_recommendations DataFrame is missing 'Property_id' column")

    # Merge collaborative recommendations with content data based on city
    new_data = collaborative_recommendations.merge(com_data, on='city',how='inner')
    new_data = new_data.drop_duplicates(subset=['name of place', 'city','type'])

    # Get content-based recommendations based on the filtered data
    content_recommendations = new_get_recommendations(city, typo,new_data)
    num_recommendations = 5

    # Limit the number of recommendations to return
    if len(content_recommendations) > num_recommendations:
        content_recommendations = content_recommendations.head(num_recommendations)

    # Ensure 'Property_id' is present in content_recommendations
    if 'Property_id' not in content_recommendations.columns:
        # Attempt to merge 'Property_id' from new_data based on common columns
        if 'name of place' in content_recommendations.columns and 'Property_id' in new_data.columns:
            content_recommendations = content_recommendations.merge(new_data[['name of place', 'Property_id']], on='name of place', how='left')
        else:
            raise ValueError("Unable to add 'Property_id' to content_recommendations. It's missing in both source DataFrames.")

    return content_recommendations
```
